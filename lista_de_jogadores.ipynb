```python
# Adicionar coluna de temporada em cada dataframe
stats_df_2021['Temporada'] = '2020/2021'
stats_df_2122['Temporada'] = '2021/2022'
stats_df_2223['Temporada'] = '2022/2023'
stats_df_2324['Temporada'] = '2023/2024'
stats_df_2425['Temporada'] = '2024/2025'

# Encontrar jogadores comuns em TODAS as temporadas
jogadores_todas_temporadas = set(stats_df_2425['Jogador'])
jogadores_todas_temporadas = jogadores_todas_temporadas.intersection(set(stats_df_2324['Jogador']))
jogadores_todas_temporadas = jogadores_todas_temporadas.intersection(set(stats_df_2223['Jogador']))
jogadores_todas_temporadas = jogadores_todas_temporadas.intersection(set(stats_df_2122['Jogador']))
jogadores_todas_temporadas = jogadores_todas_temporadas.intersection(set(stats_df_2021['Jogador']))

# Filtrar cada dataframe para manter apenas jogadores presentes em todas as temporadas
stats_df_2021_filtrado = stats_df_2021[stats_df_2021['Jogador'].isin(jogadores_todas_temporadas)]
stats_df_2122_filtrado = stats_df_2122[stats_df_2122['Jogador'].isin(jogadores_todas_temporadas)]
stats_df_2223_filtrado = stats_df_2223[stats_df_2223['Jogador'].isin(jogadores_todas_temporadas)]
stats_df_2324_filtrado = stats_df_2324[stats_df_2324['Jogador'].isin(jogadores_todas_temporadas)]
stats_df_2425_filtrado = stats_df_2425[stats_df_2425['Jogador'].isin(jogadores_todas_temporadas)]

# Concatenar todos os dataframes
todas_temporadas = pd.concat([
    stats_df_2021_filtrado,
    stats_df_2122_filtrado,
    stats_df_2223_filtrado,
    stats_df_2324_filtrado,
    stats_df_2425_filtrado
], ignore_index=True)

display(todas_temporadas.head())
print(todas_temporadas.shape)
print(f"Jogadores únicos: {todas_temporadas['Jogador'].nunique()}")
print(f"Temporadas: {todas_temporadas['Temporada'].unique()}")
print("Quantidade de jogadores por posição:")
print(todas_temporadas['Posição'].value_counts())

jogadores_unicos = todas_temporadas.drop_duplicates(subset=['Jogador'], keep='first')
contagem_posicao_unicos = jogadores_unicos['Posição'].value_counts()

print("Quantidade de jogadores únicos por posição:")
print(contagem_posicao_unicos)

contagem_posicoes = {'GK': 0, 'DF': 0, 'MF': 0, 'FW': 0}
for pos_str in jogadores_unicos['Posição'].dropna(): # .dropna() para ignorar valores nulos
    # Divide a string de posição pela vírgula para lidar com múltiplas posições
    posicoes_individuais = pos_str.split(',')
    for p in posicoes_individuais:
        if p in contagem_posicoes:
            contagem_posicoes[p] += 1

print("Quantidade de jogadores únicos por posição (GK, DF, MF, FW):")
for posicao, contagem in contagem_posicoes.items():
    print(f"{posicao}: {contagem}")
```

```python
# Fator de conversão de jardas para metros
fator_conversao = 0.9144

# Lista das colunas que estão em jardas (baseada na nossa conversa anterior)
colunas_em_jardas = [
    'Distância Média dos passes (S/ Tiros de Meta) (yd)',
    'Distância Média dos Tiros de Meta (yd)',
    'Distância Média das Ações Defensivas (yd)',
    'Distância Média das Finalizações (yd)',
    'Distância Total de Passes Completos (yd)',
    'Distância Total dos Passes Progressivos Completos (yd)',
    'Distância Total de Condução de Bola (yd)',
    'Distância Total de Condução de Bola Progressiva (yd)'
]

# Itera sobre a lista de colunas para aplicar a conversão
for coluna in colunas_em_jardas:
    if coluna in todas_temporadas.columns:
        # 1. Converte a coluna para um tipo numérico ('errors='coerce'' trata erros)
        valores_numericos = pd.to_numeric(todas_temporadas[coluna], errors='coerce')
        
        # 2. Multiplica para converter de jardas para metros
        valores_em_metros = valores_numericos * fator_conversao
        
        # 3. Arredonda para uma casa decimal e preenche possíveis nulos com 0
        todas_temporadas[coluna] = valores_em_metros.round(1).fillna(0)

# Cria um dicionário para renomear as colunas de uma só vez
mapa_renomeacao = {col: col.replace('(yd)', '(m)') for col in colunas_em_jardas}

# Renomeia as colunas no DataFrame
todas_temporadas.rename(columns=mapa_renomeacao, inplace=True)

print("Conversão de jardas para metros concluída com sucesso!")
```
